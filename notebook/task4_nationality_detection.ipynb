## 2. Testing Image Preprocessing and Validation
We can test the image validation module using a sample synthetic image.

````python
import io
import numpy as np
from PIL import Image
import sys
import os

# Add modules path
sys.path.append(os.path.abspath('Nationality_Detection'))

from modules.image_processing import validate_uploaded_file, resize_for_inference

# Create a sample in-memory JPEG image
img = Image.new('RGB', (300, 300), color=(120, 160, 200))
buf = io.BytesIO()
img.save(buf, format='JPEG')
buf.seek(0)
buf.name = 'test_portrait.jpg'

# Validate file
pil_img, bgr_img = validate_uploaded_file(buf)
print(f"Validated PIL Image Size: {pil_img.size}")
print(f"Validated BGR Array Shape: {bgr_img.shape}")
````
